```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#En caso de no poder cargar el archivo con seaborns leerlo con pandas
df= pd.read_csv("Leads_Reales_sin_valores_atipicos.csv")
```

```python
df['Mes']=df['Mes'].str.replace('Enero','1').str.replace('Febrero','2').str.replace('Marzo','3').str.replace('Abril','4').str.replace('Mayo','5').str.replace('Junio','6').str.replace('Julio','7').str.replace('Agosto','8').str.replace('Septiembre','9').str.replace('Octubre','10').str.replace('Noviembre','11').str.replace('Diciembre','12')
df
```

```python
#Graficamos todas las dispersiones entre todas las variables
sns.pairplot(df)
```

```python
#Encontramos todas las correlaciones entre las variables 
Corr_Factors=df.corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones entre las variables 
Corr_Factors1= abs(Corr_Factors)
Corr_Factors1
```

```python
#Ajustamos el mapa de calor de los coeficientes de correlación 
Heat_Map= sns.heatmap(Corr_Factors1, cmap = 'Oranges', annot=True, fmt=".2f")  
Heat_Map
```

```python
#Imprimimos el scatter plot entre la variable dependiente (total) e independiente (alcohol) 
#para observar el comportamiento en su dispersión 
from turtle import color
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=df)
```

```python
#Declaramos las variables dependientes e independientes para la regresión lineal
#Vars_Indep= df[['alcohol', 'speeding']]
Vars_Indep= df[['Efectivos']]
Var_Dep= df['Total']
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model= LinearRegression()
```

```python
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

**Modelo matemático: y= 1.63840224(x) + 39.85424212476482**


```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep,Var_Dep)
```

```python
#Predecimos los valores de total de Leads a partir de la variable "Efectivos"
#y_pred= model.predict(X=df[['alcohol', 'speeding']])
y_pred= model.predict(X=df[['Efectivos']])
y_pred
```

```python
#Insertamos la columna de predicciones en el DataFrame
df.insert(0, 'Predicciones', y_pred)
df
```

```python
#Visualizamos la gráfica comparativa entre el total real y el total predecido

sns.scatterplot(x='Efectivos', y='Total', color="blue", data=df)
sns.scatterplot(x='Efectivos', y='Predicciones', color="red", data=df)
sns.lineplot(x='Efectivos', y='Predicciones', color="red", data=df)
```

```python
#Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter=model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl=np.sqrt(coef_Deter)
coef_Correl
```
